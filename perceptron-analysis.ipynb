### Corrélation des variables numériques avec la variable cible

Dans le cadre de l'application d'un Perceptron, il est conseillé de se concentrer sur les variables les plus pertinentes, c'est-à-dire celles qui présentent une forte corrélation positive ou négative avec la cible. Les variables peu ou non corrélées (proches de 0) n'apportent généralement pas d'information utile au modèle et risquent d'ajouter du bruit.

Nous allons donc sélectionner uniquement les variables avec une corrélation absolue supérieure à 0.4, puisque les corrélations négatives sont très proche de 0.

sélecteur de variable : variance chi² et information mutuelle

```python
datacorr= data.corr()
columns_to_keep = datacorr.index[abs(datacorr['diagnosis']) >= 0.4]
filtered_data = data[columns_to_keep]
filtered_data.head()
```

Dans la matrice de corrélation, nous pouvons apercevoir des caractéristiques fortement corrélées entre elles, ce qui soulève la question de la multicolinéarité. La présence de variables redondantes peut compliquer l'entraînement du modèle, entraîner une instabilité des poids et ralentir la convergence. Pour y remédier, nous allons utiliser l'Analyse en Composantes Principales (PCA) afin de conserver l'information pertinente tout en éliminant la redondance. Cela contribue à améliorer la performance et la fiabilité du modèle.



Un perceptron est un modèle relativement simple et sensible aux dimensions inutiles. Réduire le nombre de variables aide à améliorer sa capacité à converger rapidement vers une solution optimal et réduire le risque de surapprentissage.

## Réduction de la dimensionnalité

Ici, nous séparons les données en caractéristiques (X) et variable cible (y), puis on standardise les caractéristiques pour qu'elles aient une moyenne de 0 et un écart type de 1, ce qui aide à améliorer la performance et la convergence du modèle, le perceptron est un algorithme sensible à l'échelle des données. 

```python

filtered_data.reset_index(drop=True, inplace=True)

X = filtered_data.drop(columns=['diagnosis'])
y = filtered_data['diagnosis'].reset_index(drop=True)     

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## Sélection des variables les plus pertinentes

### Sélection de features avec Boruta

Boruta est une méthode qui cherche à identifier toutes les variables utiles pour la prédiction.Boruta duplique chaque variable du dataset et mélange aléatoirement ses valeurs qui serviront de référence de bruit.
Le modèle est entraîné sur les vraies variables et les variables "bruit", puis on calcule l’importance de chaque variable. Si une variable réelle a une importance significativement supérieure à celle des variables fantômes, elle est acceptée. Si elle est inférieure, elle est rejetée.

```python
from boruta import BorutaPy
from sklearn.ensemble import RandomForestClassifier
```

```python
rf = RandomForestClassifier(n_jobs=-1, class_weight='balanced', max_depth=5, random_state=42)

boruta_selector = BorutaPy(estimator=rf, n_estimators='auto', random_state=42)

# X doit être en numpy array, pas DataFrame
boruta_selector.fit(X.values, y.values)
```

```python
selected_features = X.columns[boruta_selector.support_].to_list()
print("Features sélectionnées :", selected_features)

X_filtered = X[selected_features]
```

```python
X_filtered = X[selected_features]
```

Ici, on filtre les features pour ne pas fausser les index

```python
feature_ranks = boruta_selector.ranking_
feature_names = X.columns
accepted = boruta_selector.support_
```

### Visualisation des features importantes

On crée un dataframe des features pour leur visualisation

```python

importances_df = pd.DataFrame({
    "Feature": feature_names,
    "Rank": feature_ranks,
    "Accepted": accepted
}).sort_values("Rank")
```

```python

plt.figure(figsize=(12,6))
sns.barplot(data=importances_df, 
            x="Feature", y="Rank", 
            hue="Accepted", dodge=False, 
            palette={True: "green", False: "red"})

plt.xticks(rotation=90)
plt.title("Importance des features selon Boruta")
plt.ylabel("Rang (1 = le plus important)")
plt.xlabel("Features")
plt.show()
```

Boruta sélectionne les variables les plus pertinentes par rapport à la variable cible diagnosis en utilisant un Random Forest pour mesurer l’importance des variables. La méthode garde celles qui sont réellement utiles.
Le nombre de features est réduis et on supprime le bruit. Mais, les variables sélectionnées peuvent encore être corrélées entre elles, redondantes ou à des échelles différentes.

Nous appliquons une ACP.

### Visualisation des données en 2D (ACP)

Ici, nous appliquons l'Analyse en Composantes Principales (PCA) pour réduire les dimensions de nos données à 2 composantes principales (PC1 et PC2). Cela permet de visualiser la structure des données en deux dimensions, facilitant l'analyse des relations entre les observations.

```python
pca2 = PCA(n_components=2)
X_pca2 = pca2.fit_transform(X_scaled)

pca_df = pd.DataFrame(X_pca2, columns=['PC1', 'PC2'])
pca_df['Diagnosis'] = y.map({1: 'Malignant', 0: 'Benign'})
pca_df['Diagnosis_FR'] = pca_df['Diagnosis'].replace({'Malignant': 'Malin', 'Benign': 'Bénin'})

plt.figure(figsize=(10, 7))
sns.scatterplot(data=pca_df, x='PC1', y='PC2', hue='Diagnosis_FR', palette='coolwarm', alpha=0.7)
plt.title('Visualisation des données avec l\'ACP (2 composantes principales)')
plt.xlabel('Composante Principale 1')
plt.ylabel('Composante Principale 2')
plt.legend(title='Diagnostique')
plt.grid(True)
plt.show()
```

On observe une séparation nette entre les deux classes, "Bénin" (orange) et "Malin" (bleu), ce qui indique que les deux premières composantes capturent suffisamment d'information pour différencier les groupes. Les points bénins sont concentrés dans une région définie avec peu de dispersion, tandis que les points malins présentent une plus grande variabilité. Cette séparation claire suggère que les caractéristiques initiales contiennent des informations discriminantes pertinentes, ce qui est favorable au développement d'un modèle de classification performant.

Dans cette étape, nous appliquons l'Analyse en Composantes Principales (PCA) en spécifiant n_components=0.95, ce qui signifie que nous souhaitons conserver suffisamment de composantes principales pour expliquer au moins 95 % de la variance des données d'origine. L'utilisation de la PCA sert à réduire la dimensionnalité des données tout en préservant l'information la plus pertinente, ce qui permet d'éliminer la redondance et d'éviter les problèmes de multicolinéarité qui pourraient nuire à l'entraînement d'un modèle de classification, comme un perceptron. 

```python
pca = PCA(n_components=0.95) 
X_pca = pca.fit_transform(X_scaled)

print(f'Nombre de composantes principales retenues : {pca.n_components_}')
print('Variance expliquée par chaque composante principale :', pca.explained_variance_ratio_)
```

Les résultats montrent que six composantes principales ont été retenues lors de l'application de la PCA.

Ils suggèrent que la première composante est particulièrement importante et pourrait être utilisée pour distinguer les classes de diagnostic (bénin ou malin). En cumulant ces valeurs, on constate que les six premières composantes principales expliquent environ 89,2 % de la variance totale, ce qui montre que la majorité de l'information a été conservée malgré la réduction de dimensionnalité. Cela permet d'améliorer l'efficacité du modèle de classification que nous allons développer, tout en réduisant le risque de surajustement et en simplifiant l'apprentissage du perceptron en ne se concentrant que sur les dimensions les plus pertinentes.

## Implémentation, entraînement et évaluation du Perceptron


Nous utilisons un Perceptron monocouche car il est simple, rapide et efficace pour la classification des tumeurs. Il fonctionne bien sur des données linéairement séparables et évite le sur-apprentissage tout en offrant de bonnes performances. 

Nous séparons les données en 80 % pour l'entraînement et 20 % pour le test, afin d'évaluer la précision du modèle sur des données qu'il n'a jamais vues.

### Modèle Perceptron Sklearn

```python
X_train, X_test, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=42)
perceptron  = Perceptron(max_iter=20, random_state=42)
perceptron.fit(X_train, y_train)
y_pred = perceptron.predict(X_test)
```

 Ensuite, nous entraînons notre modèle Perceptron en définissant le paramètre max_iter à 20 itérations pour ajuster les poids. Après avoir testé différentes valeurs par paliers de dix, nous avons constaté qu'au-delà de 20 itérations, la précision atteint un maximum de 98,24 %. En dessous de cette valeur, la précision est inférieure. Puis, nous fixons random_state à 42 afin de garantir que les mêmes données sont sélectionnées à chaque exécution.

```python
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print(f"🔹 Accuracy: {accuracy:.2%}\n")
print("🔹 Classification Report:\n")
print(report)

plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=["Bénin", "Malin"], yticklabels=["Bénin", "Malin"])
plt.xlabel("Prédictions")
plt.ylabel("Valeurs réelles")
plt.title("Matrice de Confusion")
plt.show()
```

Les résultats montrent que le modèle est très efficace, avec une précision globale de 98%, ce qui signifie qu'il fait très peu d'erreurs. En regardant le rapport de classification, on voit que pour les tumeurs bénignes (0), le modèle a une précision de 97 % et un rappel de 100 %. Cela veut dire qu'il détecte tous les cas bénins et ne les confond jamais avec des tumeurs malignes. Pour les tumeurs malignes (1), la précision est parfaite (100 %), ce qui veut dire qu’il n’a jamais classé une tumeur bénigne comme maligne. Par contre, son rappel est un peu plus bas (95 %), ce qui montre que deux tumeurs malignes ont été classées à tort comme bénignes.

Globalement, le modèle fait un excellent travail, mais il manque deux cas de tumeurs malignes, ce qui peut être un problème dans un contexte médical. Pour améliorer cela, on pourrait ajuster les paramètres du modèle ou essayer d'autres méthodes comme un SVM ou une régression logistique.

### Modèle de notre classe Perceptron

Nous faisons pareil avec notre Perceptron, mais nous changeons la valeur du paramètre max_iter à 1000 qui nous donne les meilleurs résultats de performance. 

```python
X_train, X_test, y_train, y_test = train_test_split(X_pca, y, test_size=0.2, random_state=42)
y_train = y_train.reset_index(drop=True)
y_test = y_test.reset_index(drop=True)
perceptronClass = PerceptronModel(learning_rate=0.01, max_iter=1000)
perceptronClass.fit(X_train, y_train) 
y_pred = perceptronClass.predict(X_test)

accuracy, report, cm = perceptronClass.evaluate(y_test.values, y_pred) 

print(f"🔹 Accuracy: {accuracy:.2%}\n")
print("🔹 Classification Report:\n")
print(report)

plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Greens", xticklabels=["Bénin", "Malin"], yticklabels=["Bénin", "Malin"])
plt.xlabel("Prédictions")
plt.ylabel("Valeurs réelles")
plt.title("Matrice de Confusion")
plt.show()
```

Les résultats de notre modèle Perceptron montrent également une excellente performance avec une précision globale de 98%. La matrice de confusion révèle seulement deux erreurs sur 114 prédictions, avec un faux positif et un faux négatif, ce qui signifie que le modèle distingue très bien les classes. Le rapport de classification indique une précision et un rappel élevés pour les deux classes (0.99 pour "Bénin" et 0.98 pour "Malin"), assurant un bon équilibre entre faux positifs et faux négatifs. Ces performances suggèrent que le modèle généralise bien sur les données de test et qu'il n'est pas en sur-apprentissage. Pour améliorer encore le modèle, il serait possible d'ajouter plus de données ou d'expérimenter avec les hyperparamètres, mais dans l’état actuel, il offre déjà des prédictions très fiables.
